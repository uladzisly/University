```python
x = 1
s = ''
while x != 0:
  x = int(input())
  s = s+str(x)+'\n'
print(s)
cnt = 0
cnt_m = 1
s = s.split('\n')

for i in range(len(s[:-1])-1):
  if int(s[i]) == int(s[i+1]):
    cnt_m+=1
  else:
    if cnt_m>cnt:
      cnt = cnt_m
    cnt_m=1
print(cnt)


```
