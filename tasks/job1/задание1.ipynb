```python
import random

def monty_hall(i):
    stay_wins = 0
    switch_wins = 0

    for _ in range(i):
        doors = [0, 0, 1]
        random.shuffle(doors)
        chosen = random.choice([0, 1, 2])
        opened = random.choice([i for i in range(3) if i != chosen and doors[i] == 0])
        switched = next(i for i in range(3) if i != chosen and i != opened)

        if doors[chosen] == 1:
            stay_wins += 1
        if doors[switched] == 1:
            switch_wins += 1

    switch_p = switch_wins / i * 100
    stay_p = stay_wins / i * 100

    print(f"Вероятность выигрыша при смене двери: {switch_p:.2f}%".replace('.', ','))
    print(f"Вероятность выигрыша без смены двери: {stay_p:.2f}%".replace('.', ','))


monty_hall(100000)
```
