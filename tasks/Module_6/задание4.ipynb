```python
l = [int(s) for s in input().split()]
l2 = ''
ma = 0
mi = 10000
for i in l:
  if i > ma:ma=i
  if i<mi:mi=i
for x in l:
  if x == ma:l2 = l2+f'{mi} '
  elif x ==mi:l2 = l2+f'{ma} '
  else:l2 = l2+f'{(x)} '




print(l2)



```
