```python
import random

DAYS_IN_YEAR = 12 * 28

def birthday(i, people=23):
    matches = 0

    for _ in range(i):
        birthdays = [random.randint(1, DAYS_IN_YEAR) for _ in range(people)]
        if len(set(birthdays)) < people:
            matches += 1
    return matches / i * 100

result_23 = birthday(10000, 23)
print(f"Вероятность совпадения дней рождения в группе из 23 человек: {result_23:.2f}%")

result_60 = birthday(10000, 60)
print(f"Вероятность совпадения дней рождения в группе из 60 человек: {result_60:.2f}%")

print("\nЗависимость вероятности от размера группы:")
for n in range(10, 61, 5):
    print(f"{n:3} чел.: {birthday(10000, n):6.2f}%")
```
