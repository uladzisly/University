```python
from math import *
q = [float(input('Введите долготу первой точки(в градусах)')),float(input('Введите широту первой точки(в градусах)')),float(input('Введите долготу второй точки(в градусах)')),float(input('Введите широту второй точки(в градусах)'))]
cor1 = [radians(q[0]),radians(q[1])]
cor2 = [radians(q[2]),radians(q[3])]
print(cor1,cor2)

a = atan(sqrt((cos(cor2[1])*sin(cor2[0]-cor1[0]))**2 + (cos(cor1[1])*sin(cor2[1])-sin(cor1[1])*cos(cor2[1])*cos(cor2[0]-cor1[0]))**2)/sin(cor1[1])*sin(cor2[1])+cos(cor1[1])*cos(cor2[1])*cos(cor2[0]-cor1[0]))
print(a*6371.11)
```
