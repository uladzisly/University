```python
l = str(input())
l2 = ''
l3 = ''
lr = l[::-1]
s = -1
e = 1

for i in l:
    if i != 'h':
        l2 = l2+i
        s = s+1
    else:break

for x in lr:
    if x != 'h':
        l3 = x+l3
        e = e+1
    else:break

lro = l[-e-1:s+1:-1]
print(l2+'h'+lro +'h'+l3)






```
