```python
l = str(input())
l2 = ''
l3 = ''
lr = l[::-1]
s = 0
e = 2
for i in l:
  if i != 'h':
    l2 = l2+i
    s = s+1
  else:break
for x in lr:
  if x != 'h':
    l3 = x+l3
    e = e+1
  else:break
lro = l[-e:s:-1]
lro = lro.replace('h','H')
l2 = l2+'h'
l3 = 'h'+l3
print(l2+lro+l3)






```
